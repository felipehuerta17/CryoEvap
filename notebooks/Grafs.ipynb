```python
# Import modules
import CoolProp.CoolProp as CP
import numpy as np
import matplotlib.pyplot as plt
import math
from scipy import *
from scipy.integrate import simpson
# Mapas de colores
from matplotlib import cm
```

```python

Temperaturas = np.loadtxt("Temperaturas.csv", delimiter=",")
Tiempos =  np.loadtxt("Tiempos.csv", delimiter=",")
T_vapor =  np.loadtxt("T_vapor.csv", delimiter=",")
# print(Temperaturas)
dimensionless_time = Tiempos/4.8/3600
print(dimensionless_time)
```

```python
# Definir nodos 
n_x = len(Temperaturas[0])

# Radio interno
r_i = 10.24 # / m

# Espesor en pulgadas
thickness = 18

# Radio externo
r_o = r_i +thickness * 0.0254 # / m
# Espaciamiento de tiempo

# Grilla
x = np.linspace(r_i,r_o,n_x)

for i in range(len(dimensionless_time)):
    if dimensionless_time[i] == 1:
        parar = i+1

fig, ax = plt.subplots(1, 1, figsize = [5,4])
# Número de perfiles de temperatura totales a graficar
n_temp = parar
# Utilizar paleta de colores inclusiva
cividis = cm.get_cmap("cividis", n_temp)
#cividis = cm.colormaps.get_cmap("cividis", n_temp)

colour = [cividis(i/n_temp) for i in range(n_temp)]
# colours = [cividis(0), cividis(1 / 7), cividis(2 / 7), cividis(3 / 7), cividis(4 / 7), cividis

for k in range(parar):
    plt.plot(x, Temperaturas[k], color = colour[k], label = "t* = %.1f" %(dimensionless_time[k]) ) 
plt.xlabel(' $r$ / m', size = 14)
plt.ylabel(' $T$ / K', size = 14)

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid()
plt.show()
```
