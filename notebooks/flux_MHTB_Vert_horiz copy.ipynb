#### Case 1: Flux Vertical Tank
This notebook sets up the isobaric evaporation of liquid hydrogen ($LH_2$) 

```python
# Ensure that python finds the submodules
import sys
sys.path.append("..") # Adds higher directory to python modules path.

# Scientific computing
import numpy as np

# Visualisation
import matplotlib.pyplot as plt

## Module imports
# Import the storage tank Class
from cryoevap.storage_tanks import Tank

# Import Cryogen class
from cryoevap.cryogens import Cryogen
```

#### Setup tank and cryogen properties

```python
#test
d_i = 1.676
d_o = 1.676
L = 2.217

R = d_i / 2.0

# Volumen geométrico del cilindro con tapas planas
V_flat = np.pi * R**2 * L


tank_flat = Tank(
    d_i=d_i,
    d_o=d_o,
    V=V_flat,
    vapour_geometry="horizontal",
    liquid_geometry="horizontal",
    LF=0.50,
    L=L,
    head_type="flat"
)


print(f"Radio                  = {R:.10f} m")
print(f"Altura líquida         = {tank_flat.z:.10f} m")
print(f"Diferencia z - R       = {tank_flat.z - R:.3e} m")


assert np.isclose(
    tank_flat.z,
    R,
    rtol=1e-10,
    atol=1e-12
)

print("Prueba aprobada: con LF = 0.50, la altura es igual al radio.")
```

```python
d_i = 1.676
d_o = 1.676
L = 2.217

R = d_i / 2.0

# Volumen de la sección cilíndrica
V_cylinder = np.pi * R**2 * L

# Volumen de las dos tapas semiesféricas
V_heads = (4.0 / 3.0) * np.pi * R**3

# Volumen interno total
V_hemispherical = V_cylinder + V_heads


tank_hemispherical = Tank(
    d_i=d_i,
    d_o=d_o,
    V=V_hemispherical,
    vapour_geometry="horizontal",
    liquid_geometry="horizontal",
    LF=0.50,
    L=L,
    head_type="hemispherical"
)


print(f"Volumen sección cilíndrica = {V_cylinder:.10f} m³")
print(f"Volumen de las tapas       = {V_heads:.10f} m³")
print(f"Volumen total              = {V_hemispherical:.10f} m³")
print()
print(f"Radio                      = {R:.10f} m")
print(f"Altura líquida             = {tank_hemispherical.z:.10f} m")
print(f"Diferencia z - R           = {tank_hemispherical.z - R:.3e} m")


assert np.isclose(
    tank_hemispherical.z,
    R,
    rtol=1e-10,
    atol=1e-12
)

print(
    "Prueba aprobada: con LF = 0.50 y tapas "
    "semiesféricas, la altura es igual al radio."
)
```

```python
#test 3
A_T_flat_expected = (
    2.0
    * L
    * R
)

A_T_hemispherical_expected = (
    2.0
    * L
    * R
    + np.pi * R**2
)


print("ÁREA INTERFACIAL CON LF = 0.50")
print("=" * 55)

print(
    f"Plana calculada          = "
    f"{tank_flat.A_T:.10f} m²"
)

print(
    f"Plana esperada           = "
    f"{A_T_flat_expected:.10f} m²"
)

print(
    f"Semiesférica calculada   = "
    f"{tank_hemispherical.A_T:.10f} m²"
)

print(
    f"Semiesférica esperada    = "
    f"{A_T_hemispherical_expected:.10f} m²"
)


assert np.isclose(
    tank_flat.A_T,
    A_T_flat_expected,
    rtol=1e-10,
    atol=1e-12
)

assert np.isclose(
    tank_hemispherical.A_T,
    A_T_hemispherical_expected,
    rtol=1e-10,
    atol=1e-12
)

print()
print("Pruebas de A_T aprobadas.")
```

```python
R_o = d_o / 2.0

# Con d_o = d_i y LF = 0.50,
# la superficie mojada debe ser la mitad
# de la superficie total del tanque.

A_L_flat_expected = (
    np.pi * d_o * L / 2.0
    + np.pi * R_o**2
)

A_L_hemispherical_expected = (
    np.pi * d_o * L / 2.0
    + 2.0 * np.pi * R_o**2
)


print("ÁREA DE PARED EN CONTACTO CON EL LÍQUIDO")
print("=" * 60)

print(
    f"Plana calculada          = "
    f"{tank_flat.A_L:.10f} m²"
)

print(
    f"Plana esperada           = "
    f"{A_L_flat_expected:.10f} m²"
)

print(
    f"Semiesférica calculada   = "
    f"{tank_hemispherical.A_L:.10f} m²"
)

print(
    f"Semiesférica esperada    = "
    f"{A_L_hemispherical_expected:.10f} m²"
)


assert np.isclose(
    tank_flat.A_L,
    A_L_flat_expected,
    rtol=1e-10,
    atol=1e-12
)

assert np.isclose(
    tank_hemispherical.A_L,
    A_L_hemispherical_expected,
    rtol=1e-10,
    atol=1e-12
)

print()
print("Pruebas de A_L aprobadas.")
```

```python
R_o = d_o / 2.0


# ============================================================
# ÁREAS TOTALES ESPERADAS
# ============================================================

A_total_flat_expected = (
    np.pi * d_o * L
    + 2.0 * np.pi * R_o**2
)

A_total_hemispherical_expected = (
    np.pi * d_o * L
    + 4.0 * np.pi * R_o**2
)


# Con LF = 0.50, por simetría:
# A_L = A_V = A_total / 2

A_V_flat_expected = (
    A_total_flat_expected / 2.0
)

A_V_hemispherical_expected = (
    A_total_hemispherical_expected / 2.0
)


print("ÁREA DE PARED EN CONTACTO CON EL VAPOR")
print("=" * 60)

print(
    f"Plana calculada          = "
    f"{tank_flat.A_V:.10f} m²"
)

print(
    f"Plana esperada           = "
    f"{A_V_flat_expected:.10f} m²"
)

print(
    f"Semiesférica calculada   = "
    f"{tank_hemispherical.A_V:.10f} m²"
)

print(
    f"Semiesférica esperada    = "
    f"{A_V_hemispherical_expected:.10f} m²"
)


assert np.isclose(
    tank_flat.A_V,
    A_V_flat_expected,
    rtol=1e-10,
    atol=1e-12
)

assert np.isclose(
    tank_hemispherical.A_V,
    A_V_hemispherical_expected,
    rtol=1e-10,
    atol=1e-12
)


# ============================================================
# COMPROBACIÓN DEL ÁREA TOTAL
# ============================================================

assert np.isclose(
    tank_flat.A_L + tank_flat.A_V,
    A_total_flat_expected,
    rtol=1e-10,
    atol=1e-12
)

assert np.isclose(
    (
        tank_hemispherical.A_L
        + tank_hemispherical.A_V
    ),
    A_total_hemispherical_expected,
    rtol=1e-10,
    atol=1e-12
)


print()
print("Pruebas de A_V aprobadas.")
print("También se comprobó que A_L + A_V = A_total.")
```

```python
# Parámetros exclusivamente para la prueba numérica
P_test = 101325.0       # Pa
T_air_test = 298.15    # K
U_L_test = 0.0045      # W/(m² K)
U_V_test = 0.0045      # W/(m² K)
eta_w_test = 0.0


for tank in (
    tank_flat,
    tank_hemispherical
):

    # Configuración del fluido
    tank.cryogen = Cryogen(
        name="Hydrogen"
    )

    tank.cryogen.set_coolprops(
        p=P_test
    )

    # Configuración térmica del tanque
    tank.set_HeatTransProps(
        U_L=U_L_test,
        U_V=U_V_test,
        T_air=T_air_test,
        Q_b_fixed=0.0,
        Q_roof=0.0,
        eta_w=eta_w_test
    )
```

```python
for name, tank in (
    ("Flat", tank_flat),
    ("Hemispherical", tank_hemispherical)
):

    print(name)
    print("  U_roof =", tank.U_roof)
    print("  U_L =", tank.U_L)
    print("  U_V =", tank.U_V)
    print("  T_air =", tank.T_air)
    print("  Q_b_fixed =", tank.Q_b_fixed)
    print("  Cryogen =", tank.cryogen.name)
    print("  T_sat =", tank.cryogen.T_sat)
    print("  k_V_avg =", tank.cryogen.k_V_avg)
    print()
```

```python
t_f = 6*3600 
tank_flat.evaporate(t_f)
tank_hemispherical.evaporate(t_f)
```

```python
print(
    "Flat success:",
    tank_flat.sol.success
)

print(
    "Hemispherical success:",
    tank_hemispherical.sol.success
)

print()
print(tank_flat.data.keys())
print(tank_hemispherical.data.keys())
```

```python
# ============================================================
# TANQUE CON TAPAS PLANAS
# ============================================================

z_flat = np.asarray(
    tank_flat.data["z"]
)

A_T_flat = np.asarray(
    tank_flat.data["A_T"]
)

half_chord_flat = np.sqrt(
    np.maximum(
        tank_flat.d_i * z_flat
        - z_flat**2,
        0.0
    )
)

A_T_flat_expected = (
    2.0
    * tank_flat.L
    * half_chord_flat
)

error_flat = np.max(
    np.abs(
        A_T_flat
        - A_T_flat_expected
    )
)


# ============================================================
# TANQUE CON TAPAS SEMIESFÉRICAS
# ============================================================

z_hemi = np.asarray(
    tank_hemispherical.data["z"]
)

A_T_hemi = np.asarray(
    tank_hemispherical.data["A_T"]
)

half_chord_hemi = np.sqrt(
    np.maximum(
        tank_hemispherical.d_i * z_hemi
        - z_hemi**2,
        0.0
    )
)

A_T_hemi_expected = (
    2.0
    * tank_hemispherical.L
    * half_chord_hemi
    + np.pi
    * half_chord_hemi**2
)

error_hemi = np.max(
    np.abs(
        A_T_hemi
        - A_T_hemi_expected
    )
)


print("VALIDACIÓN TEMPORAL DE A_T")
print("=" * 55)

print(
    f"Error máximo tapas planas: "
    f"{error_flat:.3e} m²"
)

print(
    f"Error máximo tapas semiesféricas: "
    f"{error_hemi:.3e} m²"
)


assert np.allclose(
    A_T_flat,
    A_T_flat_expected,
    rtol=1e-10,
    atol=1e-12
)

assert np.allclose(
    A_T_hemi,
    A_T_hemi_expected,
    rtol=1e-10,
    atol=1e-12
)

assert len(A_T_flat) == len(
    tank_flat.data["Time"]
)

assert len(A_T_hemi) == len(
    tank_hemispherical.data["Time"]
)

print()
print(
    "Reconstrucción temporal de A_T aprobada."
)
```
