
**Introducción**

En el presente archivo se presenta un modelo 1-D de evaporación de líquido criogénico en un regimen isobárico. Se adaptaron las ecuaciones del modelo de Huerta & Vesovic (2019) correspondiente a estanques cilíndricos verticales, modificando la geometría de las ecuaciones para modelar el comportamiento en un estanque horizontal.
A continuación, se muestran los parámetros del modelo:

**1. Geometría del modelo**

En primer lugar, se presentan las ecuaciones que modelan la física del problema. A continuación, se muestra una imagen del estanque, visto desde el lado.
Como se puede ver, existe una fase líquida y una fase vapor. Para cada una de las fases, se modela un ingreso de calor por separado. Además existe una fuente de calor extra proveniente del piso del estanque. 
En el almacenamiento, el calor provoca que cierta parte del líquido se evapore. Para mantener presiones seguras dentro del estanque y asegurar regimen isobárico, un flujo de vapor se libera en la parte superior del estanque como Boil Off Gas (BOG).


![Vista desde el lado](vista_frontal.png)

El estanque modelado cuenta con tapas planas. Es decir, en las tapas del estanque se cuenta con circunferencias. La vista desde esa perspectiva se puede ver a continuación. 

![Vista tapas](vista_tapas.png)

Como se puede apreciar en la figura, la circunferencia cuenta con fase líquida y fase vapor. Es importante destacar el ángulo alpha, que se forma por el diámetro de la circunferencia en 90°, y la altura del líquido. A medida que aumenta la altura del líquido, este ángulo será mayor. 

A continuación, se muestran las ecuaciones que representan el modelo geométrico de ambas fases:

**Area del liquido y del vapor en las tapas circulares (1)**

Para determinar las áreas ocupadas por el líquido y el vapor en las tapas circulares del estanque, se analiza la sección transversal del cilindro. Dado un nivel de llenado h medido desde la base, la superficie cubierta por el líquido corresponde al área de un segmento circular. Esta área depende directamente de la geometría del cilindro y del ángulo subtendido por el nivel de líquido. A partir de este segmento se obtiene también el área correspondiente al vapor, calculada como la parte no ocupada del círculo.

$$
A_{L}^{\text{tapa}}(h) \;=\; \,\big(R^2arcos(\frac{R-h}{R})- (R-h)\cdot \sqrt{2Rh-h^2} ),\\
A_{V}^{\text{tapa}}(h) = \pi R^{2}-A_{L}^{\text{transv}}(h)
$$

**Volumen del líquido (2)**

Una vez determinada el área del líquido en la sección transversal del cilindro, el volumen ocupado por la fase líquida se obtiene directamente al extender dicha área a lo largo del largo del estanque. De esta forma, el volumen del líquido resulta de multiplicar el área del segmento circular correspondiente al nivel de llenado por la longitud 
L.

$$

V_{L}^{\text{}}(h) \;=\; L\,\big(R^2arcos(\frac{R-h}{R})- (R-h)\cdot \sqrt{2Rh-h^2} )

$$

El volumen del vapor, en cambio, se calcula como la diferencia entre el volumen total del estanque y el volumen del líquido previamente determinado.

$$
V_{V}=L\big(\pi R^{2}-A_{L}^{\text{transv}}(h)\big) 
$$

**Perimetros de liquido y vapor en las tapas (3)**

El contacto entre las fases y la pared del estanque en las tapas circulares se describe mediante los perímetros mojados por el líquido y el vapor. Estos perímetros dependen del ángulo central asociado al nivel de llenado, ya que este ángulo determina qué fracción de la circunferencia está cubierta por cada fase. De esta forma, tanto el perímetro mojado por el líquido como el correspondiente al vapor pueden expresarse en función del ángulo geométrico 
𝜃


$$
s_{L}(h)=2R\,\theta(h),\quad
s_{V}(h)=2\pi R - 2R\,\theta(h)
$$

$$
\theta(h) \;=\; \,\arccos\!\left(\frac{R-h}{R}\right)
$$

**Áreas de pared mojadas por cada fase (4)**

La superficie interior del estanque que está en contacto con el líquido o con el vapor se determina combinando el aporte de las tapas circulares y el de la superficie lateral del cilindro. El área mojada de cada fase resulta entonces del perímetro mojado correspondiente, multiplicado por la longitud del estanque, más el área ocupada por dicha fase en ambas tapas. Esta descomposición permite evaluar con precisión la transferencia de calor hacia cada fase.

$$
A_{w,L}=L\,s_{L}(h)+2\cdot A_{L}^{\text{tapa}}(h) ,\qquad
A_{w,V}=L\,s_{V}(h)+ 2\cdot A_{V}^{\text{tapa}}(h)
$$

**Largo de cuerda en la interfase LV (5)**

El largo de cuerda describe la extensión de la interfase líquido–vapor cuando se observa la sección transversal del cilindro. Esta magnitud corresponde a la línea recta que conecta los dos puntos donde el nivel de líquido intersecta la circunferencia. Su valor depende directamente del nivel 
ℎ


$$
c(h)\;=\; \; 2\sqrt{\,2Rh-h^2\,},\\
$$

**Área interfacial líquido–vapor (6)**

El área interfacial corresponde a la superficie total de contacto entre el líquido y el vapor dentro del estanque. Geométricamente, esta área se obtiene al extender a lo largo del cilindro el largo de la cuerda de la interfase determinada en la sección transversal. Su valor depende directamente del nivel de llenado, ya que este define la forma y extensión de la línea de contacto entre ambas fases.

$$
A_{i}(h)=2L\sqrt{\,2Rh-h^2 \,}
$$

**Derivada del área interfacial líquido-vapor (7)**

La variación del área interfacial con respecto al nivel de llenado es una magnitud clave en el balance de energía, ya que describe cómo cambia la superficie disponible para transferencia entre las fases. La derivada se obtiene aplicando la regla de la cadena a la expresión del área interfacial, lo que permite evaluar de forma explícita cómo pequeñas variaciones en 
ℎ
h afectan la extensión de la interfase.

$$
\frac{\mathrm{d}A_i}{\mathrm{d}h}
= \frac{\mathrm{d}}{\mathrm{d}h}\big[2L\sqrt{2Rh - h^2}\big]
= \frac{2L(R - h)}{\sqrt{2Rh - h^2}}.
$$

**2. Balances de masa y energía**

A continuación, se muestra el modelo termodinámico. Se muestran los distintos balances de masa y energía que componen el modelo.

**Balance de masa del líquido (8)**

El balance de masa en la fase líquida relaciona la variación temporal del volumen de líquido con la evaporación que ocurre en la interfase. Como la densidad del líquido se considera constante, la disminución de su masa es directamente proporcional al caudal másico evaporado desde esta fase. De esta forma, el balance refleja cómo la evaporación reduce la cantidad de líquido en el estanque.

$$
\frac{\mathrm{d}}{\mathrm{d}t}\big(\rho_{L}V_{L}\big)=-\dot B_{L}.
$$

**Balance global de masa y extracción de BOG (9)**

En condiciones isobáricas, el vapor generado por la evaporación no puede acumularse indefinidamente en la fase gaseosa; para mantener la presión constante, debe extraerse un caudal de gas (BOG). El balance global de masa —que considera tanto líquido como vapor— relaciona entonces la evaporación, la acumulación en el espacio de vapor y el caudal extraído. Este balance permite determinar, en cada instante, el flujo de BOG requerido para conservar la presión del tanque.

$$
-\dot B=\frac{\mathrm{d}}{\mathrm{d}t}\big(\rho_{L}V_{L}+\rho_{V}V_{V}\big).
$$

**Balance de energía del líquido (10)**

El balance de energía en la fase líquida considera todas las fuentes de calor que ingresan al líquido dentro del estanque. Este calor proviene de la pared en contacto con el líquido (lateral y tapas), de la losa o apoyos inferiores, y del vapor que transfiere energía adicional en la interfase. La energía recibida se reparte entre dos procesos: el aumento de la entalpía sensible del líquido y el aporte necesario para evaporar parte de la masa, mediante el calor de vaporización. De esta forma, el balance vincula los flujos de calor con el cambio energético total de la fase líquida.

$$
U_{s,L}\,A_{w,L}\,[T_{\infty}-T_{L}]
+U_{e,L}\,2A_{L}^{\text{tapa}}(h)\,[T_{\infty}-T_{L}]
+\dot Q_{\text{slab}}(h)\\
\qquad+\dot Q_{VL}-\dot B_{L}\,h_{V}(T_{L})
=\frac{\mathrm{d}}{\mathrm{d}t}\big[\rho_{L}V_{L}\,h_{L}(T_{L})\big].
$$

**3. Modelo de transferencia de calor**

**Flujo interfacial vapor a líquido (conducción) (11)**

La transferencia de calor entre el vapor y el líquido ocurre a través de la interfase, donde ambas fases no están en equilibrio térmico. En este modelo, se asume que el vapor es más caliente que el líquido, por lo que el flujo de calor se dirige hacia la fase líquida. Este flujo se calcula aplicando la ley de Fourier a una superficie plana, usando el gradiente vertical de temperatura del vapor evaluado en la posición de la interfase. El resultado representa el calor conducido desde el vapor hacia el líquido en cada instante.

$$
\dot Q_{VL}(t)=-\,k_{V}\,A_{i}(h)\,
\left.\frac{\partial T_{V}}{\partial z}\right|_{z=h}.
$$

**Ecuación de energía para el vapor (vertical) (12)**

La evolución de la temperatura del vapor dentro del estanque se describe mediante una ecuación de balance energético unidimensional en la dirección vertical. En este balance intervienen tres mecanismos principales: (i) la difusión térmica debida a conducción, (ii) la advección generada por el movimiento ascendente del vapor producido en la interfase, y (iii) un término fuente volumétrico que representa el calentamiento proveniente de la pared del tanque, modelado mediante un coeficiente global de transferencia de calor. Adicionalmente, la variación del área transversal disponible para el vapor con la altura introduce un término geométrico adicional que corrige la difusión efectiva. Este balance permite calcular el perfil de temperatura del vapor en función de la altura y del tiempo.

$$
\frac{\partial T_V}{\partial t}
= \alpha_V\!\left[
\frac{\partial^2 T_V}{\partial z^2}
+ \frac{A'(z)}{A(z)}\,\frac{\partial T_V}{\partial z}
\right]
- v_z(z)\,\frac{\partial T_V}{\partial z}
+ \frac{\dot{s}_{\mathrm{wall}}(z,T_V)}{\rho_V\,c_{p,V}} \, .
$$

**Velocidad media de advección en el vapor (13)**

El movimiento del vapor dentro del estanque se aproxima mediante una velocidad vertical promedio, consistente con una formulación unidimensional del flujo. Esta velocidad surge del caudal másico de vapor generado en la interfase y se distribuye sobre el área transversal ocupada por el vapor. De este modo, la expresión relaciona directamente la evaporación con el transporte vertical de energía dentro de la fase gaseosa.

$$
v_{z}(h)=\frac{\dot B_{L}/\rho_{V}}{A_{T}^{\text{}}(h)}.
$$

**Condiciones de borde para la PDE del vapor (14)**

Para resolver la ecuación de energía del vapor es necesario especificar condiciones de borde físicamente consistentes. En la interfase líquido–vapor, se impone continuidad de temperatura, ya que ambas fases se encuentran en equilibrio térmico local. En la parte superior del estanque, en cambio, se asume una condición adiabática, lo que implica ausencia de flujo de calor hacia el exterior por esa superficie. Estas dos condiciones permiten cerrar el problema bien planteado para la evolución del perfil de temperatura del vapor.

$$
T_{V}(t,0)=T_{L}(t),\qquad
\left.\frac{\partial T_{V}}{\partial z}\right|_{z=l_{V}^{\text{vert}}(t)}=0.
$$

**Calor por pared hacia el líquido (15)**

El líquido recibe calor desde todas las porciones de la pared del estanque que se encuentran en contacto con él, tanto en la superficie lateral como en las tapas circulares sumergidas. El flujo neto depende del coeficiente global de transferencia de calor correspondiente a cada región y del área mojada calculada previamente. Esta expresión cuantifica la contribución térmica desde la pared hacia la fase líquida, que posteriormente participa en el balance de energía del líquido.

$$
\dot Q_{L}
=U_{s,L}\,L\,s_{L}(h)\,[T_{\infty}-T_{L}]
+U_{e,L}\,2A_{L}^{\text{cap}}(h)\,[T_{\infty}-T_{L}].
$$

**Calor por pared hacia el vapor (16)**

El vapor también recibe calor desde las superficies interiores del estanque que están en contacto con esta fase. Este aporte incluye tanto la porción lateral expuesta al vapor como las tapas superiores no sumergidas. El flujo neto depende del coeficiente global de transferencia de calor correspondiente a cada región y de las áreas mojadas por el vapor. Esta expresión permite cuantificar la energía transferida desde la pared hacia el vapor, la cual influye directamente en su calentamiento y en la dinámica térmica del espacio gaseoso.

$$
\dot Q_{V}
=U_{s,V}\,L\,s_{V}(h)\,[T_{\infty}-\bar T_{V}(t)]
+U_{e,V}\,2\big(\pi R^{2}-A_{L}(h)\big)\,[T_{\infty}-\bar T_{V}(t)].
$$

**Relación de la extracción total (17)**

La extracción total de BOG se obtiene considerando dos contribuciones: la masa evaporada desde la fase líquida y la acumulación (o disminución) de masa en el espacio de vapor. Si el vapor se expande o contrae debido a cambios térmicos, su masa varía en el tiempo, afectando la cantidad neta que debe extraerse para mantener la presión constante. Esta expresión vincula la evaporación y la dinámica del vapor con el caudal efectivo de gas que abandona el tanque.

$$
\dot B=\dot B_{L}-\frac{\mathrm{d}}{\mathrm{d}t}\big(\rho_{V}V_{V}\big).
$$
