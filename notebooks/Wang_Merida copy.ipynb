#### Case 1: Isobaric evaporation of liquid hydrogen in a 18.09 $m^3$ Cylindrical tank experiment MHTB.

This notebook sets up the isobaric evaporation of liquid hydrogen ($LH_2$) 

```python
# Rutas e importaciones
import sys
sys.path.append("..")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from cryoevap.storage_tanks import Tank
from cryoevap.cryogens import Cryogen
```

```python
# Operación
P = 101_300.0                 # Pa
T_air_C = 15.0                # °C
T_air = T_air_C + 273.15      # K

# Geometría de Wang y Mérida
V_tank = 4.89                 # m³
d_i = 1.676                   # m
d_o = d_i                     # m
L_tank = 2.217                # m

# Caso de la Figura 12
LF_initial = 0.30
q_target = 0.50               # W/m²
eta_w_baseline = 0.97

# Horizontes temporales
short_hours = 5.0
long_hours = 160.0

# Discretización
short_dz_target = 0.01        # m
short_output_interval = 60    # s
long_dz_target = 0.01         # m
long_output_interval = 300    # s

# Controles de ejecución
RUN_SHORT_SIMULATIONS = True
RUN_LONG_SIMULATIONS = False
RUN_OPTIONAL_ETA_DIAGNOSTIC = False

assert P > 0.0
assert T_air > 0.0
assert 0.0 < LF_initial < 1.0
assert 0.95 <= eta_w_baseline <= 0.99
assert short_dz_target > 0.0 and long_dz_target > 0.0
```

```python
# Propiedades de saturación y U equivalente
hydrogen_reference = Cryogen(name="hydrogen")
hydrogen_reference.set_coolprops(P)

T_sat = hydrogen_reference.T_sat
delta_T_initial = T_air - T_sat
U_equivalent = q_target / delta_T_initial

radius = d_i / 2.0
V_geometry_flat_heads = np.pi * radius**2 * L_tank
A_lateral_flat_heads = np.pi * d_i * L_tank
A_ends_flat_heads = 2.0 * np.pi * radius**2
A_total_flat_heads = A_lateral_flat_heads + A_ends_flat_heads

assert delta_T_initial > 0.0
assert U_equivalent > 0.0
assert np.isclose(V_geometry_flat_heads, V_tank, rtol=5e-4)

print("CASO BASE")
print("=" * 65)
print(f"Presión                         = {P/1000:.3f} kPa")
print(f"Temperatura ambiente            = {T_air_C:.2f} °C ({T_air:.2f} K)")
print(f"Temperatura de saturación       = {T_sat:.8f} K")
print(f"Flux líquido objetivo           = {q_target:.6f} W/m²")
print(f"U equivalente                   = {U_equivalent:.10f} W/(m² K)")
print(f"Nivel líquido inicial           = {100*LF_initial:.2f} %")
print(f"eta_w basal                     = {eta_w_baseline:.4f}")
print(f"Volumen con tapas planas        = {V_geometry_flat_heads:.8f} m³")
print(f"Área externa con tapas planas   = {A_total_flat_heads:.8f} m²")
```

```python
def build_tank(orientation, eta_w=eta_w_baseline):
    """Construye un tanque independiente para una orientación dada."""

    cryogen = Cryogen(name="hydrogen")
    cryogen.set_coolprops(P)

    if orientation == "vertical":
        tank = Tank(
            d_i=d_i,
            d_o=d_o,
            V=V_tank,
            vapour_geometry="cylindrical",
            liquid_geometry="cylindrical",
            LF=LF_initial,
        )
        Q_b_fixed = None

    elif orientation == "horizontal":
        tank = Tank(
            d_i=d_i,
            d_o=d_o,
            V=V_tank,
            vapour_geometry="horizontal",
            liquid_geometry="horizontal",
            LF=LF_initial,
            L=L_tank,
        )
        Q_b_fixed = 0.0

    else:
        raise ValueError("orientation debe ser 'vertical' o 'horizontal'.")

    tank.cryogen = cryogen
    tank.set_HeatTransProps(
        U_L=U_equivalent,
        U_V=U_equivalent,
        T_air=T_air,
        Q_b_fixed=Q_b_fixed,
        Q_roof=0.0,
        eta_w=eta_w,
    )
    return tank


def postprocess_mass_balance(tank):
    """Reconstruye inventarios, evaporación y BOG desde balances de masa."""

    time_h = np.asarray(tank.sol.t, dtype=float) / 3600.0
    V_L = np.asarray(tank.data["V_L"], dtype=float)
    LF = np.asarray(tank.data["LF"], dtype=float)
    Tv_avg = np.asarray(tank.data["Tv_avg"], dtype=float)
    rho_V_avg = np.asarray(tank.data["rho_V_avg"], dtype=float)

    m_L = tank.cryogen.rho_L * V_L
    m_V = rho_V_avg * (V_tank - V_L)

    dMLdt = np.gradient(m_L, time_h)
    dMVdt = np.gradient(m_V, time_h)
    evaporation = -dMLdt
    BOG = -(dMLdt + dMVdt)

    # Promedio diagnóstico basado en Tv_avg. No reemplaza una integración
    # local sobre la pared cuando se disponga del campo térmico completo.
    q_vapour_mean_from_Tvavg = U_equivalent * (T_air - Tv_avg)
    q_liquid = np.full_like(time_h, U_equivalent * (T_air - T_sat))

    results = pd.DataFrame({
        "time_h": time_h,
        "Tv_avg_K": Tv_avg,
        "LF": LF,
        "LF_pct": 100.0 * LF,
        "mL_kg": m_L,
        "mV_kg": m_V,
        "dMLdt_kg_h": dMLdt,
        "evap_kg_h": evaporation,
        "BOG_kg_h": BOG,
        "q_liquid_W_m2": q_liquid,
        "q_vapour_mean_from_Tvavg_W_m2": q_vapour_mean_from_Tvavg,
    })

    delta_total_mass = (m_L[-1] + m_V[-1]) - (m_L[0] + m_V[0])
    released_mass = np.trapezoid(BOG, time_h)
    mass_balance_error = released_mass + delta_total_mass

    # Consistencia de la derivada almacenada, excluyendo t=0 porque el
    # valor interno se inicializa en cero.
    dMLdt_internal = tank.cryogen.rho_L * np.asarray(tank.data["dV_L"]) * 3600.0
    derivative_error = np.max(np.abs(dMLdt[1:] - dMLdt_internal[1:]))

    assert tank.sol.success
    assert np.all(np.isfinite(results.select_dtypes(include=[np.number]).to_numpy()))
    assert np.all((results["LF"] > 0.0) & (results["LF"] <= 1.0))
    assert abs(mass_balance_error) < 1e-8
    assert derivative_error < 1e-5

    diagnostics = {
        "delta_total_mass_kg": delta_total_mass,
        "released_mass_kg": released_mass,
        "mass_balance_error_kg": mass_balance_error,
        "max_dMLdt_error_excluding_t0_kg_h": derivative_error,
    }

    return results, diagnostics


def simulate_case(
    orientation,
    simulation_hours,
    dz_target,
    output_interval_seconds,
    eta_w=eta_w_baseline,
):
    """Ejecuta una simulación nueva y devuelve tanque, resultados y malla."""

    tank = build_tank(orientation=orientation, eta_w=eta_w)

    n_z = max(3, 1 + int(np.round(tank.l_V / dz_target)))
    tank.z_grid = np.linspace(0.0, 1.0, n_z)
    tank.time_interval = int(output_interval_seconds)

    dz_physical = tank.l_V / (n_z - 1)
    simulation_seconds = int(simulation_hours * 3600.0)

    print(
        f"Ejecutando {orientation}: eta_w={eta_w:.3f}, "
        f"t={simulation_hours:.1f} h, nodos={n_z}, "
        f"dz≈{dz_physical:.5f} m"
    )

    tank.evaporate(simulation_seconds)
    results, diagnostics = postprocess_mass_balance(tank)

    mesh = {
        "orientation": orientation,
        "eta_w": eta_w,
        "n_z": n_z,
        "dz_physical_m": dz_physical,
        "output_interval_s": int(output_interval_seconds),
        "nfev": tank.sol.nfev,
    }

    return tank, results, diagnostics, mesh
```

```python
def extract_curve(dataframe, time_column, value_column):
    """Extrae, limpia y ordena una curva temporal del Excel."""

    missing = {time_column, value_column}.difference(dataframe.columns)
    if missing:
        raise KeyError(f"Faltan columnas: {sorted(missing)}")

    curve = dataframe[[time_column, value_column]].copy()
    curve.columns = ["time", "value"]
    curve["time"] = pd.to_numeric(curve["time"], errors="coerce")
    curve["value"] = pd.to_numeric(curve["value"], errors="coerce")
    curve = curve.dropna().sort_values("time").reset_index(drop=True)

    assert not curve.empty
    assert np.all(np.isfinite(curve[["time", "value"]].to_numpy()))
    return curve


sheet_T = wm_sheets["12a_T"]
Tl_wm = extract_curve(sheet_T, "t_Tl_h", "Tl_K")
Tv_vert_wm = extract_curve(sheet_T, "t_Tv_vert_h", "Tv_vert_K")
Tv_horiz_wm = extract_curve(sheet_T, "t_Tv_horiz_h", "Tv_horiz_K")

sheet_LF = wm_sheets["12b_LF"]
LF_wm = extract_curve(sheet_LF, "t_common_h", "LF_common_pct")

sheet_mL = wm_sheets["12c_mL"]
mL_wm = extract_curve(sheet_mL, "t_common_h", "mL_common_kg")

sheet_dMLdt = wm_sheets["12d_dMLdt"]
dMLdt_vert_wm = extract_curve(sheet_dMLdt, "t_vert_h", "dMLdt_vert_kg_h")
dMLdt_horiz_wm = extract_curve(sheet_dMLdt, "t_horiz_h", "dMLdt_horiz_kg_h")

sheet_BOG = wm_sheets["12e_BOG"]
BOG_vert_wm = extract_curve(sheet_BOG, "t_vert_h", "BOG_vert_kg_h")
BOG_horiz_wm = extract_curve(sheet_BOG, "t_horiz_h", "BOG_horiz_kg_h")

print("Curvas digitalizadas cargadas correctamente.")
```

```python
# FUNCIÓN DE SENSIBILIDAD DE eta_w: TANQUE VERTICAL

def simulate_vertical_eta(eta_value):
    """
    Ejecuta una simulación vertical de cinco horas para un valor
    determinado de eta_w.

    Se mantienen constantes U_L, U_V, geometría, presión,
    nivel inicial y discretización.
    """

    # Crear un criógeno independiente
    hydrogen_case = Cryogen(name="hydrogen")
    hydrogen_case.set_coolprops(P)

    # Crear un tanque nuevo para cada simulación
    tank = Tank(
        d_i=d_i,
        d_o=d_o,
        V=V_tank,
        vapour_geometry="cylindrical",
        liquid_geometry="cylindrical",
        LF=LF_initial
    )

    tank.cryogen = hydrogen_case

    tank.set_HeatTransProps(
        U_L=U_equivalent,
        U_V=U_equivalent,
        T_air=T_air,
        Q_b_fixed=None,
        Q_roof=0.0,
        eta_w=eta_value
    )

    # Mantener aproximadamente el mismo espaciamiento espacial
    n_z = max(
        3,
        1 + int(
            np.round(
                tank.l_V / dz_target
            )
        )
    )

    tank.z_grid = np.linspace(
        0.0,
        1.0,
        n_z
    )

    tank.time_interval = output_interval

    # Ejecutar cinco horas
    tank.evaporate(t_short_seconds)

    if not tank.sol.success:
        raise RuntimeError(
            f"La simulación falló para eta_w={eta_value}."
        )

    # Tiempo en horas
    time_h = tank.sol.t / 3600.0

    # Inventarios
    V_L = tank.data["V_L"]

    m_L = (
        tank.cryogen.rho_L
        * V_L
    )

    m_V = (
        tank.data["rho_V_avg"]
        * (
            V_tank - V_L
        )
    )

    # Tasas obtenidas mediante balances de masa
    dMLdt = np.gradient(
        m_L,
        time_h
    )

    dMVdt = np.gradient(
        m_V,
        time_h
    )

    evaporation = -dMLdt

    BOG = -(
        dMLdt
        + dMVdt
    )

    # Comprobaciones mínimas
    variables = [
        tank.data["Tv_avg"],
        tank.data["LF"],
        m_L,
        m_V,
        evaporation,
        BOG
    ]

    for values in variables:
        if not np.all(np.isfinite(values)):
            raise ValueError(
                f"Se encontraron valores inválidos "
                f"para eta_w={eta_value}."
            )

    return {
        "eta_w": eta_value,
        "Tv_final": tank.data["Tv_avg"][-1],
        "delta_Tv_final": (
            tank.data["Tv_avg"][-1]
            - tank.cryogen.T_sat
        ),
        "LF_final": tank.data["LF"][-1],
        "mL_final": m_L[-1],
        "evap_final": evaporation[-1],
        "BOG_final": BOG[-1],
        "evap_min": np.min(evaporation),
        "BOG_min": np.min(BOG),
        "nfev": tank.sol.nfev
    }
```

```python
# ============================================================
# SENSIBILIDAD GRUESA DE eta_w
# ============================================================

eta_values_coarse = [
    0.30,
    0.50,
    0.70,
    0.85,
    0.97
]

eta_sensitivity_vertical = []

for eta_value in eta_values_coarse:

    print(
        f"Ejecutando eta_w = {eta_value:.2f}..."
    )

    result = simulate_vertical_eta(
        eta_value
    )

    eta_sensitivity_vertical.append(
        result
    )

print("Sensibilidad terminada.")
```

```python
# FUNCIÓN PARA EXTRAER CURVAS DESDE UN DATAFRAME

def extract_curve(
    dataframe,
    time_column,
    value_column
):
    """
    Extrae y limpia una curva temporal desde un DataFrame.

    Parameters
    ----------
    dataframe : pandas.DataFrame
        Pestaña del Excel.
    time_column : str
        Nombre de la columna temporal.
    value_column : str
        Nombre de la columna con la variable.

    Returns
    -------
    pandas.DataFrame
        DataFrame con columnas 'time' y 'value'.
    """

    curve = dataframe[
        [
            time_column,
            value_column
        ]
    ].copy()

    curve.columns = [
        "time",
        "value"
    ]

    curve["time"] = pd.to_numeric(
        curve["time"],
        errors="coerce"
    )

    curve["value"] = pd.to_numeric(
        curve["value"],
        errors="coerce"
    )

    curve = curve.dropna(
        subset=[
            "time",
            "value"
        ]
    )

    curve = curve.sort_values(
        by="time"
    )

    curve = curve.reset_index(
        drop=True
    )

    return curve
```

```python
# FUNCIÓN PARA SIMULACIONES DE LARGO PLAZO

def simulate_long_case(
    orientation,
    eta_value,
    simulation_hours=160.0,
    output_interval_seconds=300
):
    """
    Simula un tanque vertical u horizontal y devuelve sus
    resultados temporales en un DataFrame.

    Parameters
    ----------
    orientation : str
        "vertical" o "horizontal".
    eta_value : float
        Fracción eta_w.
    simulation_hours : float
        Duración total de la simulación en horas.
    output_interval_seconds : int
        Intervalo entre resultados almacenados.

    Returns
    -------
    tank : Tank
        Objeto simulado.
    results : pandas.DataFrame
        Resultados temporales.
    """

    hydrogen_case = Cryogen(name="hydrogen")
    hydrogen_case.set_coolprops(P)

    if orientation == "vertical":

        tank = Tank(
            d_i=d_i,
            d_o=d_o,
            V=V_tank,
            vapour_geometry="cylindrical",
            liquid_geometry="cylindrical",
            LF=LF_initial
        )

        Q_b_fixed = None

    elif orientation == "horizontal":

        tank = Tank(
            d_i=d_i,
            d_o=d_o,
            V=V_tank,
            vapour_geometry="horizontal",
            liquid_geometry="horizontal",
            LF=LF_initial,
            L=L_tank
        )

        Q_b_fixed = 0.0

    else:
        raise ValueError(
            "orientation debe ser 'vertical' o 'horizontal'."
        )

    tank.cryogen = hydrogen_case

    tank.set_HeatTransProps(
        U_L=U_equivalent,
        U_V=U_equivalent,
        T_air=T_air,
        Q_b_fixed=Q_b_fixed,
        Q_roof=0.0,
        eta_w=eta_value
    )

    # Malla espacial con dz aproximado de 1 cm
    n_z = max(
        3,
        1 + int(
            np.round(
                tank.l_V / dz_target
            )
        )
    )

    tank.z_grid = np.linspace(
        0.0,
        1.0,
        n_z
    )

    tank.time_interval = output_interval_seconds

    simulation_seconds = int(
        simulation_hours * 3600
    )

    print(
        f"Ejecutando {orientation}, "
        f"eta_w={eta_value:.2f}, "
        f"t={simulation_hours:.1f} h..."
    )

    tank.evaporate(simulation_seconds)

    if not tank.sol.success:
        raise RuntimeError(
            f"Falló la simulación {orientation}, "
            f"eta_w={eta_value}."
        )

    time_h = tank.sol.t / 3600.0

    V_L = np.asarray(
        tank.data["V_L"]
    )

    LF = np.asarray(
        tank.data["LF"]
    )

    Tv_avg = np.asarray(
        tank.data["Tv_avg"]
    )

    rho_V_avg = np.asarray(
        tank.data["rho_V_avg"]
    )

    m_L = tank.cryogen.rho_L * V_L

    m_V = (
        rho_V_avg
        * (V_tank - V_L)
    )

    # Tasas mediante balance de masa
    dMLdt = np.gradient(
        m_L,
        time_h
    )

    dMVdt = np.gradient(
        m_V,
        time_h
    )

    evaporation = -dMLdt

    BOG = -(
        dMLdt
        + dMVdt
    )

    results = pd.DataFrame({
        "time_h": time_h,
        "Tv_avg_K": Tv_avg,
        "LF": LF,
        "LF_pct": 100.0 * LF,
        "mL_kg": m_L,
        "mV_kg": m_V,
        "dMLdt_kg_h": dMLdt,
        "evap_kg_h": evaporation,
        "BOG_kg_h": BOG
    })

    # Comprobaciones mínimas
    assert np.all(
        np.isfinite(
            results.select_dtypes(
                include=[np.number]
            ).to_numpy()
        )
    )

    assert np.all(
        results["LF"] > 0.0
    )

    assert np.all(
        results["mL_kg"] > 0.0
    )

    print(
        f"Terminada: {orientation}, "
        f"eta_w={eta_value:.2f}, "
        f"nfev={tank.sol.nfev}"
    )

    return tank, results
```
